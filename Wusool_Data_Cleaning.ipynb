```python
#import library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the data
wusool_users = pd.read_excel('wusool_users_2023.xlsx')
```

```python
wusool_users.head()
```

```python
#use .shape
wusool_users.shape
```

```python
# Check the total count of missing values
wusool_users.isnull().sum()
```

```python
# Check the percentage of missing values
wusool_users.isna().sum() / len(wusool_users)
```

```python
# Visualizing Missing Data
plt.figure(figsize=(10,6))
sns.displot(
 wusool_users.isna().melt(value_name="missing"),
    y="variable",
    hue="missing",
    multiple="fill",
    aspect=1.25)
```

```python
# dropping rows with missing values
wusool_users = wusool_users.dropna().reset_index(drop=True)
```

```python
# Check the total count of missing values
wusool_users.isnull().sum()
```

```python
# Step 3: Save the DataFrame back to an Excel file
wusool_users.to_excel('output_file.xlsx', index=False)
```
