```python
#import library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the data
qurrah_users = pd.read_excel('qurrah_users_2023.xlsx')
```

```python
qurrah_users.head()
```

```python
#use .shape
qurrah_users.shape
```

```python
#Checking missing values stats
qurrah_users.info()
```

```python
# Check the total count of missing values
qurrah_users.isnull().sum()
```

```python
# Check the percentage of missing values
qurrah_users.isna().sum() / len(qurrah_users)
```

```python
# Visualizing Missing Data
plt.figure(figsize=(10,6))
sns.displot(
 qurrah_users.isna().melt(value_name="missing"),
    y="variable",
    hue="missing",
    multiple="fill",
    aspect=1.25)
```

```python
# dropping rows with missing values
qurrah_users = qurrah_users.dropna().reset_index(drop=True)
```

```python
# Check the total count of missing values
qurrah_users.isnull().sum()
```

```python
# Step 3: Save the DataFrame back to an Excel file
qurrah_users.to_excel('output_file.xlsx', index=False)
```
